```python
# Import necessary libraries
import logging
import sys

from awsglue.transforms import *  # AWS Glue built-in transformations
from awsglue.utils import getResolvedOptions  # Utility to retrieve job parameters
from pyspark.context import SparkContext  # Entry point to Spark functionality
from pyspark.sql import functions as F
from awsglue.context import GlueContext  # AWS Glue-specific context extending Spark
from awsglue.job import Job  # AWS Glue Job class for managing job lifecycle
from awsgluedq.transforms import EvaluateDataQuality  # Optional: For data quality checks
from awsglue import DynamicFrame  # AWS Glue DynamicFrame abstraction
import gs_derived  # Custom transformation script/module (ensure this is included in your Glue script dependencies)

logging.basicConfig(level=logging.INFO, format="%(asctime)s %(levelname)s %(message)s")
logger = logging.getLogger(__name__)
```

```python
# Define a helper function to run Spark SQL on Glue DynamicFrames
def sparkSqlQuery(glueContext, query, mapping, transformation_ctx) -> DynamicFrame:
    """
    Executes a Spark SQL query on a mapping of DynamicFrames and returns the result as a new DynamicFrame.

    Parameters:
        glueContext (GlueContext): The Glue context object.
        query (str): The SQL query to execute.
        mapping (dict): A dictionary where keys are table aliases and values are DynamicFrames.
        transformation_ctx (str): The transformation context name (used for job bookmarks and debugging).

    Returns:
        DynamicFrame: The result of the SQL query as a DynamicFrame.
    """
    for alias, frame in mapping.items():
        # Register each DynamicFrame as a temporary view for Spark SQL
        frame.toDF().createOrReplaceTempView(alias)

    # Execute SQL query using Spark
    result = spark.sql(query)

    # Convert resulting DataFrame back to a Glue DynamicFrame
    return DynamicFrame.fromDF(result, glueContext, transformation_ctx)

```

```python
# Retrieve job parameters passed at runtime
args = getResolvedOptions(sys.argv, ['JOB_NAME', 'S3_BUCKET_NAME'])
output_bucket = args['S3_BUCKET_NAME']
output_prefix = 'cleaned_data'

# Initialize Spark and Glue contexts
sc = SparkContext()
glueContext = GlueContext(sc)
spark = glueContext.spark_session

# Initialize the Glue job with the provided JOB_NAME
job = Job(glueContext)
job.init(args['JOB_NAME'], args)
logger.info("Starting Glue ETL job %s; output bucket=%s", args['JOB_NAME'], output_bucket)

DEFAULT_DATA_QUALITY_RULESET = """
    Rules = [
        ColumnCount > 0
    ]
"""


def validate_and_deduplicate(frame, frame_name, key_columns, required_columns):
    data_frame = frame.toDF()
    missing_columns = sorted(set(required_columns).difference(data_frame.columns))
    if missing_columns:
        raise ValueError(f"{frame_name} is missing required columns: {missing_columns}")

    null_counts = data_frame.select(*[
        F.sum(F.col(column).isNull().cast("long")).alias(column)
        for column in required_columns
    ]).first().asDict()
    invalid_counts = {
        column: count or 0
        for column, count in null_counts.items()
        if count
    }
    if invalid_counts:
        raise ValueError(f"{frame_name} has nulls in required columns: {invalid_counts}")

    before_count = data_frame.count()
    if key_columns:
        data_frame = data_frame.dropDuplicates(key_columns)
    else:
        data_frame = data_frame.dropDuplicates()
    duplicate_count = before_count - data_frame.count()
    logger.info("%s: removed %d duplicate rows", frame_name, duplicate_count)
    return DynamicFrame.fromDF(data_frame, glueContext, f"{frame_name}_validated")
```

```python
# Normalize the sales dates and combine the three years without duplicate column names.
SqlQuery1 = '''
SELECT TO_DATE(orderdate, 'MM/dd/yyyy') AS orderdate,
       TO_DATE(stockdate, 'MM/dd/yyyy') AS stockdate,
       ordernumber, productkey, customerkey, territorykey,
       orderlineitem, orderquantity, 2015 AS sales_year
FROM aw_sales_2015
WHERE orderdate RLIKE '^[0-9]{2}/[0-9]{2}/[0-9]{4}$'
  AND stockdate RLIKE '^[0-9]{2}/[0-9]{2}/[0-9]{4}$'
  AND TO_DATE(orderdate, 'MM/dd/yyyy') IS NOT NULL
  AND TO_DATE(stockdate, 'MM/dd/yyyy') IS NOT NULL
UNION ALL
SELECT TO_DATE(orderdate, 'MM/dd/yyyy') AS orderdate,
       TO_DATE(stockdate, 'MM/dd/yyyy') AS stockdate,
       ordernumber, productkey, customerkey, territorykey,
       orderlineitem, orderquantity, 2016 AS sales_year
FROM aw_sales_2016
WHERE orderdate RLIKE '^[0-9]{2}/[0-9]{2}/[0-9]{4}$'
  AND stockdate RLIKE '^[0-9]{2}/[0-9]{2}/[0-9]{4}$'
  AND TO_DATE(orderdate, 'MM/dd/yyyy') IS NOT NULL
  AND TO_DATE(stockdate, 'MM/dd/yyyy') IS NOT NULL
UNION ALL
SELECT TO_DATE(orderdate, 'MM/dd/yyyy') AS orderdate,
       TO_DATE(stockdate, 'MM/dd/yyyy') AS stockdate,
       ordernumber, productkey, customerkey, territorykey,
       orderlineitem, orderquantity, 2017 AS sales_year
FROM aw_sales_2017
WHERE orderdate RLIKE '^[0-9]{2}/[0-9]{2}/[0-9]{4}$'
  AND stockdate RLIKE '^[0-9]{2}/[0-9]{2}/[0-9]{4}$'
  AND TO_DATE(orderdate, 'MM/dd/yyyy') IS NOT NULL
  AND TO_DATE(stockdate, 'MM/dd/yyyy') IS NOT NULL
'''

orderdatestockdatedate_node1743763366006 = sparkSqlQuery(
    glueContext,
    query=SqlQuery1,
    mapping={
        "aw_sales_2015": aw_sales_2015_node1743763329845,
        "aw_sales_2016": aw_sales_2016_node1743764120348,
        "aw_sales_2017": aw_sales_2017_node1743829333023
    },
    transformation_ctx="orderdatestockdatedate_node1743763366006"
)
```

```python
# ==============================
# EXPORTING TO S3 WITH QUALITY CHECKS
# ==============================

exports = [
    (longint_node1743761564692, "aw_product_subcategories", "ExporttoS3_node1743761671860", ["productsubcategorykey"], ["productsubcategorykey", "subcategoryname", "productcategorykey"]),
    (longint_node1743762226688, "aw_products", "ExporttoS3_node1743762306649", ["productkey"], ["productkey"]),
    (productcategory_to_int_node1743746865206, "aw_product_category", "ExporttoS3_node1743746915045", ["productcategorykey"], ["productcategorykey", "categoryname"]),
    (longint_node1743765434879, "aw_territories", "ExportS3_node1743765449831", ["salesterritorykey"], ["salesterritorykey"]),
    (longint_node1743762747732, "aw_returns", "ExporttoS3_node1743763059295", [], ["returndate", "territorykey", "productkey", "returnquantity"]),
    (longintremoveorderlineitemorderquantity_node1743763724020, "aw_sales", "ExporttoS3_node1743763820156", ["ordernumber", "orderlineitem", "sales_year"], ["orderdate", "stockdate", "ordernumber", "productkey", "customerkey", "orderlineitem", "orderquantity", "sales_year"]),
    (TotalChildrenint_node1743760550959, "aw_customers", "ExporttoS3_node1743677309923", ["customerkey"], ["customerkey", "birthdate"])
]

for frame, folder, node, key_columns, required_columns in exports:
    try:
        frame = validate_and_deduplicate(frame, folder, key_columns, required_columns)
        EvaluateDataQuality().process_rows(
            frame=frame,
            ruleset=DEFAULT_DATA_QUALITY_RULESET,
            publishing_options={
                "dataQualityEvaluationContext": f"EvaluateDataQuality_{node}",
                "enableDataQualityResultsPublishing": True
            },
            additional_options={
                "dataQualityResultsPublishing.strategy": "BEST_EFFORT",
                "observations.scope": "ALL"
            }
        )
        output_path = f"s3://{output_bucket}/{output_prefix}/{folder}/"
        glueContext.write_dynamic_frame.from_options(
            frame=frame,
            connection_type="s3",
            format="glueparquet",
            connection_options={"path": output_path, "partitionKeys": []},
            format_options={"compression": "snappy"},
            transformation_ctx=node
        )
        logger.info("Wrote %s to %s", folder, output_path)
    except Exception:
        logger.exception("Failed processing export %s", folder)
        raise

job.commit()
logger.info("Glue ETL job %s completed successfully", args['JOB_NAME'])
```
